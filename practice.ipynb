```python
'''
1.温度数据分析
某城市一周最高气温(摄氏度)为[28, 30, 29, 31, 32, 30, 29]
- 计算平均气温、最高气温和最低气温
- 找出气温超过30摄氏度的天数
'''
```

```python
import numpy as np
day = np.array([28, 30, 29, 31, 32, 30, 29])
print(f'平均气温：{np.mean(day)}, 最高气温：{np.max(day)}, 最低气温：{np.min(day)}')
print(f'气温超过30摄氏度的天数：{len(day[day > 30])}')
```
