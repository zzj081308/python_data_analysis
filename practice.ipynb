```python
from jedi.inference.compiled import access

'''
1.温度数据分析
某城市一周最高气温(摄氏度)为[28, 30, 29, 31, 32, 30, 29]
- 计算平均气温、最高气温和最低气温
- 找出气温超过30摄氏度的天数
'''
```

```python
import numpy as np
temps = np.array([28, 30, 29, 31, 32, 30, 29])
print(f'平均气温：{np.mean(temps)}, 最高气温：{np.max(temps)}, 最低气温：{np.min(temps)}')
print(f'气温超过30摄氏度的天数：{len(temps[temps > 30])}')
print(f'气温超过30摄氏度的天数：{np.cumsum(np.where(temps > 30, 1, 0))[-1]}')
print(f'气温超过30摄氏度的天数：{np.count_nonzero(temps > 30)}')
```

```python
"""
2.学生成绩统计
某班级5名学生数学成绩分别为[85, 90, 78, 92, 88]
- 计算成绩的平均分、中位数和标准差
- 将成绩转换为百分制(满分为100分)
"""
```

```python
scores = np.array([85, 90, 78, 92, 88])
print(f'平均分:{np.average(scores)}')
print(f'中位数:{np.median(scores)}')
print(f'标准差:{np.std(scores)}')
```
