# 一、 ndarray
## 1.ndarray的特性
### 多维性


```python
import numpy as np
```

```python
arr = np.array(5)   #创建零维ndarray数组，即常量
print(arr)
print(f'arr的维度：{arr.ndim}')     #获取维度
```

```python
arr = np.array([1, 2, 3])   #创建零维ndarray数组，即常量
print(arr)
print(f'arr的维度：{arr.ndim}')     #获取维度
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])   #创建零维ndarray数组，即常量
print(arr)
print(f'arr的维度：{arr.ndim}')     #获取维度
```

### 同质性

```python
arr = np.array([1, 'hello'])    #不同数据类型会自动转换为相同类型
print(arr)
```

```python
arr = np.array([1, 2.5])    #不同数据类型会自动转换为相同类型
print(arr)
```

## 2.ndarray的属性

```python
arr = np.array(1)
print(arr)
print(f'arr的形状：{arr.shape}')    #几行几列
print(f'arr的维度：{arr.ndim}')
print(f'元素个数：{arr.size}')
print(f'元素类型：{arr.dtype}')
print(f'元素转置：{arr.T}')
print(f'单元素占用字节数：{arr.itemsize}')
print(f'数组总占用字节数：{arr.nbytes}')
print(f'内存存储方式(是否连续)：\n{arr.flags}')
```

```python
arr = np.array([1, 2, 3])
print(arr)
print(f'arr的形状：{arr.shape}')
print(f'arr的维度：{arr.ndim}')
print(f'元素个数：{arr.size}')
print(f'元素类型：{arr.dtype}')
print(f'元素转置：{arr.T}')
print(f'单元素占用字节数：{arr.itemsize}')
print(f'数组总占用字节数：{arr.nbytes}')
print(f'内存存储方式(是否连续)：\n{arr.flags}')
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print(f'arr的形状：{arr.shape}')
print(f'arr的维度：{arr.ndim}')
print(f'元素个数：{arr.size}')
print(f'元素类型：{arr.dtype}')
print(f'元素转置：\n{arr.T}')
print(f'单元素占用字节数：{arr.itemsize}')
print(f'数组总占用字节数：{arr.nbytes}')
print(f'内存存储方式(是否连续)：\n{arr.flags}')
```

## 3.ndarray的创建

```python
# 基础创建方法
list = [1, 2, 3]
arr = np.array(list, dtype=np.float64)    #其实是传入一个列表,还可以指定类型
print(arr)
```

```python
# copy方法
arr1 = np.copy(arr)
print(arr1)
```

### 预定义形状

```python
# 全0
arr = np.zeros((2, 3))  #传入一个shape类型，(元组),创建一个全0的2行3列矩阵
print(arr)
print(f'默认类型：{arr.dtype}')
```

```python
# 全1
arr = np.ones((2, 3))
print(arr)
```

```python
# 未初始化
arr = np.empty((3, 3))
print(arr)
```

```python
# 固定值
arr = np.full((2, 3), 5)
print(arr)
```

```python
# 模仿
# 最新一次的arr是2行3列的矩阵
arr2 = np.zeros_like(arr)   #模仿arr的形状，生成全0
print(arr2)
arr2 = np.ones_like(arr)    #模仿arr的形状，生成全1
print(arr2)
arr2 = np.full_like(arr, 8) #模仿arr的形状，生成全8
print(arr2)
arr2 = np.empty_like(arr)   #模仿arr的形状，生成全随机
print(arr2)
```

```python
# 按步长
arr = np.arange(0, 10, 2)   #和range类似，左闭右开，步长为2
print(arr)
```

```python
# 按数量取区间等间距
arr = np.linspace(0, 100, 5)     #此处左右闭合，等间距取5个数
print(arr)
```

```python
# 对数步长
arr = np.logspace(0, 4, 3, base=2)
# 0  1  2  3  4
# ^     ^     ^
# 0到4之间取3个数，基底为2
#log2(1) = 0, log2(4) = 2, log2(16) = 4
print(arr)
# 其实最终得到的是等比数列
```

```python
# 再演示一个
arr = np.logspace(1, 5, 5, base=10)
print(arr)
#得到的就是1，10，100，1000，10000
```

```python
# 特殊矩阵
# 单位矩阵：主对角线上数字为1，其他数字为0
arr = np.eye(3, 4, dtype=int)
print(arr)
```

```python
# 对角矩阵：主对角线非0，其他数字为0
arr = np.diag([1, 2, 3])
print(arr)
```

```python
# 随机数组生成
# 生成0，1之间的随机浮点数
arr = np.random.rand(2, 3)
print(arr)
```

```python
# 生成区间随机浮点数
arr = np.random.uniform(1, 100, (2, 3))
print(arr)
```

```python
# 生成区间随机整数数
arr = np.random.randint(1, 100, (2, 3))
print(arr)
```

```python
# 生成区间随机浮点数 （正态分布）
arr = np.random.randn(2, 3)
print(arr)
```

```python
# 设置随机数种子
np.random.seed(20)
arr = np.random.randint(1, 10, (2, 3))
print(arr)
```

## 5.索引与切片

### 一维数组的索引与切片

```python
# 生成20个1~100随机整数
arr = np.random.randint(1, 100, 20)
print(arr)
```

```python
print(arr[0], arr[1], arr[2])
```

```python
print(arr[:3])  #前3个
print(arr[3:])  #3个之后的所有
```

```python
print(arr[:-3])     #除去后3个
print(arr[-3:])     #后3个
```

```python
print(arr[3:5])     #下标3~5之间两个，左闭右开
```

```python
print(arr[5:3])     #反向索引为空
```

```python
print(arr[(arr > 50) & (arr < 80)])     #大于50小于80的元素
```

```python
print(arr[slice(3, 5)])     #等价[3:5]
print(arr[slice(0, 20, 2)])     #0~20个元素之间，步长为2
```

### 二维数组的索引与切片

```python
arr = np.random.randint(1, 100, (5, 5))     #1到100的5行5列的随机整数
print(arr)
```

```python
print(arr[2, 2])    #下标2，2
```

```python
# 下标2行，前3列
print(arr[:2, :3],"\n")
# 下标2行，后3列
print(arr[1, 2:])
```

```python
# 下标1行，后4位
print(arr[1, -4:])
print(arr[1, 1:][arr[1, 1:] > 50])
```

```python
# 取列
print(arr[:, 2])
print(arr[:, 2][arr[:, 2] > 50])
```
